```python
import pandas as pd
import numpy as np

# 1. Load raw dataset (checks raw_churn.csv first, or falls back to the uploaded sample file)
import os
file_path = "raw_churn.csv" if os.path.exists("raw_churn.csv") else "customer_churn_sample (3).csv"
df = pd.read_csv(file_path)

print("=== FILE LOADED ===", file_path)
print("=== SHAPE & COLUMNS ===")
print(df.shape)
print(df.dtypes)

print("\n=== MISSING VALUES ===")
print(df.isnull().sum())

print("\n=== DUPLICATE ROWS ===")
print("Exact duplicates:", df.duplicated().sum())

# 2. Clean data
clean_df = df.copy()

# Strip extra spaces from text columns
for col in clean_df.select_dtypes(include="object").columns:
    clean_df[col] = clean_df[col].astype(str).str.strip()

# Drop exact duplicate rows
clean_df = clean_df.drop_duplicates()

# Fill numeric missing values with median, categorical missing values with mode
for col in clean_df.columns:
    if clean_df[col].isnull().sum() > 0:
        if pd.api.types.is_numeric_dtype(clean_df[col]):
            clean_df[col] = clean_df[col].fillna(clean_df[col].median())
        else:
            clean_df[col] = clean_df[col].fillna(clean_df[col].mode()[0])

print("\n=== AFTER CLEANING: MISSING & DUPLICATES ===")
print("Remaining nulls:", clean_df.isnull().sum().sum())
print("Remaining duplicates:", clean_df.duplicated().sum())

print("\n=== SUMMARY STATISTICS (NUMERIC) ===")
print(clean_df.describe())

# 3. Export cleaned CSV
clean_df.to_csv("cleaned_churn.csv", index=False)
print("\nSaved cleaned_churn.csv successfully!")
```
